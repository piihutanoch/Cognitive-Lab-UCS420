```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) Add 2°C to every reading
temperature = temperature + 2
print("Updated temperatures:", temperature)

# b) Convert Celsius to Fahrenheit
fahrenheit = (9/5) * temperature + 32
print("Fahrenheit:", fahrenheit)

# c) Readings greater than 32°C using Boolean indexing
greater_than_32 = temperature[temperature > 32]
print("Readings > 32°C:", greater_than_32)

# d) Count readings exceeding 32°C
count = np.sum(temperature > 32)
print("Count:", count)

# e) Vectorization performs operations on the entire array at once,
# making the code shorter, faster, and more efficient than using a loop.
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a) Total steps recorded
total = np.sum(steps)
print("Total steps:", total)

# b) Mean number of steps
mean = np.mean(steps)
print("Mean steps:", mean)

# c) Maximum and minimum values
maximum = np.max(steps)
minimum = np.min(steps)
print("Maximum:", maximum)
print("Minimum:", minimum)

# d) Total steps for each day (axis=0)
day_totals = np.sum(steps, axis=0)
print("Total steps for each day:", day_totals)

# e) Total steps for each user (axis=1)
user_totals = np.sum(steps, axis=1)
print("Total steps for each user:", user_totals)

# f) User/day position containing maximum steps
max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum position (user, day):", max_position)
print("Maximum steps:", steps[max_position])
```

```python
import numpy as np

# a) Create the original NumPy array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

# b) Slice elements from index 1 to 4
subset = original[1:5]
print("Subset:", subset)

# c) Modify first element of subset
subset[0] = 999
print("Original after modifying subset:", original)
print("Subset:", subset)

# d) Create another slice using copy()
copied = original[1:5].copy()
copied[0] = 500
print("Original after modifying copied:", original)
print("Copied array:", copied)

# e) Create 1 to 12 and reshape into 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)
print("3 x 4 Matrix:\n", matrix)

# f) NumPy indexing and slicing
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:\n", matrix[0:2, 1:3])

# g) Flatten using flatten() and ravel()
flat = matrix.flatten()
raveled = matrix.ravel()
print("Flatten:", flat)
print("Ravel:", raveled)

# h) Modify ravel() result
raveled[0] = 100
print("After modifying ravel:")
print("Original matrix:\n", matrix)
print("Ravel:", raveled)

# i) Modify flatten() result
flat[0] = 200
print("After modifying flatten:")
print("Original matrix:\n", matrix)
print("Flatten:", flat)

# j) Shape, dimensions, size, and type
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
import numpy as np

# Given data
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a) Shape and dimensions of X
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# b) X.T and explanation
XT = X.T
print("\nX.T:\n", XT)
# Transpose converts rows into columns and columns into rows.

# c) Matrix product X.T @ X
XTX = XT @ X
print("\nX.T @ X:\n", XTX)

# d) Inverse of X.T @ X
XTX_inv = np.linalg.inv(XTX)
print("\nInverse of X.T @ X:\n", XTX_inv)

# e) Ordinary Least Squares:
# β = (X.T X)^(-1) X.T y
beta = XTX_inv @ XT @ y
print("\nOLS coefficients (β):", beta)

# f) Coefficients
print("\nCoefficients:")
print("Sleep hours coefficient:", beta[0])
print("Activity level coefficient:", beta[1])
print("Stress level coefficient:", beta[2])

# g) Predict assistance score for new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("\nPredicted assistance score:", predicted_score)
```
