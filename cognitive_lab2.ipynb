```python
import random
random.seed(1024170373)

a= [random.randint(100, 900) for _ in range(100)]
a
```

```python
odd= [x for x in a if x%2 !=0]
print(odd)
print("Count of odd numbers is "+ str(len(odd)))
```

```python
even= [x for x in a if x%2==0]
print(even)
print("Count of even numbers is "+ str(len(even)))
```

```python
pn=[
    i for i in a
    if i>1 and all(i%j!=0 for j in range(2, i))
]

print(pn)
print("Count of prime numbers is "+ str(len(pn)))
```

```python
num=0
maxcount=0

for i in a:
  count= a.count(i)
  if(count>maxcount):
    maxcount= count
    num= i

print("the number that occurs most frequenty is " + str(num) + " and it occurs " + str(maxcount) + " times")
```
